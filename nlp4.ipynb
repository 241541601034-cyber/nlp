```python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
nltk.download('stopwords')
nltk.download('punkt')
text = "This is a sample sentence showing stopword removal."
stop_words = set(stopwords.words('english'))
tokens = word_tokenize(text.lower())
filtered_tokens = [word for word in tokens if word not in stop_words]
print("Original:", tokens)
print("Filtered:", filtered_tokens)
```

```python
import nltk
from nltk.tokenize import word_tokenize

nltk.download('punkt')

lemmatization_dict = {
    'running': 'run',
    'ran': 'run',
    'children': 'child',
    'games': 'game'
}

def lemmatize(text):
    tokens = word_tokenize(text)
    return [lemmatization_dict.get(token.lower(), token) for token in tokens]

text = "The children are running games and ran yesterday"

lemmas = lemmatize(text)

print("Lemmatization Results:")
print(lemmas)
```

```python
import nltk
from nltk.tokenize import word_tokenize
import re

nltk.download('punkt')

lemmatization_rules = {
    'dogs': 'dog',
    'ran': 'run',
    'cat': 'cat',
    'running': 'run'
}

def lemmatize_word(word):
    return lemmatization_rules.get(word, word)

def lemmatize_text(text):
    text = re.sub(r'[^\w\s]', ' ', text.lower())
    tokens = word_tokenize(text)
    return ' '.join(lemmatize_word(word) for word in tokens)

input_text = "the dogs are running"
lemmatized_text = lemmatize_text(input_text)

print("Lemmatized text:", lemmatized_text)
```
